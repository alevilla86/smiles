```python
# predict_nn_original.py
import os, json, pickle
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import load_model

def _ensure_features(df, features):
    missing = [c for c in features if c not in df.columns]
    if missing:
        raise ValueError(f"Faltan columnas requeridas: {missing}")
    return df[features]

def load_artifacts(artifacts_dir):
    model = load_model(os.path.join(artifacts_dir, "model.keras"))
    with open(os.path.join(artifacts_dir, "x_scaler.pkl"), "rb") as f:
        x_scaler = pickle.load(f)
    with open(os.path.join(artifacts_dir, "meta.json"), "r") as f:
        meta = json.load(f)
    return model, x_scaler, meta

def predict_from_dataframe(df: pd.DataFrame, artifacts_dir: str):
    model, x_scaler, meta = load_artifacts(artifacts_dir)
    features = meta["features"]
    y_min = meta.get("y_min", None)
    y_max = meta.get("y_max", None)
    clip_output = meta.get("clip_output", False)

    X = _ensure_features(df, features).to_numpy(dtype=float)
    Xs = x_scaler.transform(X)
    y_pred = model.predict(Xs, verbose=0).ravel()
    if clip_output and (y_min is not None) and (y_max is not None):
        y_pred = np.clip(y_pred, y_min, y_max)
    return y_pred

def predict_from_csv(csv_path: str, artifacts_dir: str, save_csv: str = None):
    df = pd.read_csv(csv_path)
    y_pred = predict_from_dataframe(df, artifacts_dir)
    if save_csv:
        out = df.copy()
        out["pred"] = y_pred
        out.to_csv(save_csv, index=False)
        print(f"✅ Predicciones guardadas en: {save_csv}")
    return y_pred

if __name__ == "__main__":
    import argparse
    parser = argparse.ArgumentParser(description="Inferencia NN original")
    parser.add_argument("--artifacts", type=str, default="artifacts_nn_original", help="Carpeta con model.keras, x_scaler.pkl, meta.json")
    parser.add_argument("--csv", type=str, default="", help="CSV de entrada para predecir")
    parser.add_argument("--save_csv", type=str, default="", help="Ruta de salida con las predicciones")
    args = parser.parse_args()

    if args.csv:
        preds = predict_from_csv(args.csv, args.artifacts, save_csv=args.save_csv or None)
        print(f"Predicciones (n={len(preds)}):\n{preds[:10]}{' ...' if len(preds)>10 else ''}")
    else:
        print("Provee --csv para predecir un archivo.")
```
